Select a GPU runtime. This hosts synthesis; your local application retains story extraction, quote modes and mastering.
Install the pinned package, upload the three Python files named below from your checkout, and run the final cell. It prints a new public URL to put in local-test.settings.json (endpoint) or .env (OMNIVOICE_URL).
